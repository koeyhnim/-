# **1.환경준비**

* 라이브러리 Import

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split              # 데이터 분할
from sklearn.metrics import *                                     # 평가
from sklearn.preprocessing import StandardScaler, MinMaxScaler    # 스케일링
```

# **2.Regression : Advertising**

## (1) 데이터 전처리

### 1) 데이터 준비

```
path = 'https://raw.githubusercontent.com/DA4BAM/dataset/master/advertising.csv'
data1 = pd.read_csv(path)
data1.head()
```

```
# x, y 나누기
target = 'Sales'
x = data1.drop(target, axis=1)
y = data1.loc[:, target]
```

### 2) 가변수화

### 3) 데이터분할

```
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size = .3  , random_state = 20)
```

### 4) Scaling
* 필요할 때만 스케일링 : KNN, SVM, K-Means, Deep Learning

```
scaler = MinMaxScaler()
x_train_s = scaler.fit_transform(x_train)
x_val_s = scaler.transform(x_val)
```

## (2) 모델링
* 필요한 함수들 불러오기
* 모델 선언
* 학습
* 예측
* 성능 검증

```
# LinearRegression으로 모델링을 수행합니다.
from sklearn.linear_model import LinearRegression
```

### 1) 모델 선언

```
model = LinearRegression()
```

### 2) 학습

```
model.fit(x_train_s, y_train)
```

### 3) 검증 : 예측

```
pred = model.predict(x_val_s)
pred
```

### 4) 검증 : 평가
만든 모델은 얼마나 정확한지 검증해 봅시다.



```
mean_absolute_error(y_val, pred)
```

# **3.Classification : 대학원 지원**

## (1) 데이터 전처리

### 1) 데이터 준비

```
path = "https://raw.githubusercontent.com/DA4BAM/dataset/master/Graduate_apply.csv"
data2 = pd.read_csv(path)
data2.head()
```

```
# x, y 분할
target = 'admit'
x = data2.drop(target, axis=1)
y = data2.loc[:, target]
```

### 2) 가변수화

```
cat_cols = ['rank']
x = pd.get_dummies(x, columns = cat_cols, drop_first = True)
```

### 3) 데이터분할

```
x_train, x_val, y_train, y_val = train_test_split(x, y, test_size = .3, random_state = 20)
```

### 4) Scaling
* 필요할 때만 스케일링 : KNN, SVM, K-Means, Deep Learning

```
scaler = MinMaxScaler()
x_train_s = scaler.fit_transform(x_train)
x_val_s = scaler.transform(x_val)
```

## (2) 모델링
* 필요한 함수들 불러오기
* 모델 선언
* 학습
* 예측
* 성능 검증

```
# DecisionTreeClassifier로 모델링을 수행합니다.
from sklearn.tree import DecisionTreeClassifier
```

### 1) 모델 선언

```
model = DecisionTreeClassifier()
```

### 2) 학습

```
model.fit(x_train_s, y_train)
```

### 3) 검증 : 예측

```
pred = model.predict(x_val_s)
pred
```

### 4) 검증 : 평가
만든 모델은 얼마나 정확한지 검증해 봅시다.



```
accuracy_score(y_val, pred)
```
